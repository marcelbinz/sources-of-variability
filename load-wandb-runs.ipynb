```python
import wandb
import pandas as pd
```

# ITC

## Level 1

```python
# load filtered runs from project

api = wandb.Api()

runs_conditions = api.runs(
    "mirkothalmann-helmholtz-munich/source-of-variablity",
    filters={
        "config.dataset": "itc_full",
        "config.d_model": 32,
        "config.d_ff": 256,
        "config.more_shuffle": "nothing",
        "config.rnd_seed": 2
    }
)
```

```python
l_dfs_condition = []
for run in runs_conditions:
    df_history = run.history()
    df_history.columns = [run.name + c for c in df_history.columns]
    l_dfs_condition.append(df_history)
df_conditions = pd.concat(l_dfs_condition[0:4], axis=1).reset_index()
df_conditions.rename(columns={"index":"epoch"}, inplace=True)
df_conditions.to_csv("wandb/wandb_export_itc_conditions_code.csv", index=False)
```

## Masklength

```python
# load filtered runs from project

api = wandb.Api()

# first all runs with varied masklength
runs_mask = api.runs(
    "mirkothalmann-helmholtz-munich/source-of-variablity",
    filters={
        "config.dataset": "itc_full",
        "config.d_model": 32,
        "config.d_ff": 256,
        "config.condition": "original",
        "config.more_shuffle": "nothing",
        "config.masktype":"windowed_causal",
        "config.windowsize": {"$in": [1, 3, 10, 15, 30, 40]}
    }
)

# then upper baseline, i.e., original data set
runs_upper = api.runs(
    "mirkothalmann-helmholtz-munich/source-of-variablity",
    filters={
        "config.dataset": "itc_full",
        "config.d_model": 32,
        "config.d_ff": 256,
        "config.condition": "original",
        "config.more_shuffle": "nothing",
        "config.masktype":"causal",
        "config.rnd_seed":2
    }
)

# then lower baseline, i.e., ID no history
runs_lower = api.runs(
    "mirkothalmann-helmholtz-munich/source-of-variablity",
    filters={
        "config.dataset": "itc_full",
        "config.d_model": 32,
        "config.d_ff": 256,
        "config.condition": "id_nohist",
        "config.more_shuffle": "nothing",
        "config.masktype":"causal",
        "config.rnd_seed":2
    }
)
```

```python
l_dfs_mask = []
l_dfs_upper = []
l_dfs_lower = []

for run in runs_mask:
    df_masklength = run.history()
    df_masklength.columns = [run.name + c for c in df_masklength.columns]
    l_dfs_mask.append(df_masklength)
for run in runs_upper:
    df_upper = run.history()
    df_upper.columns = [run.name + c for c in df_upper.columns]
    l_dfs_upper.append(df_upper)
for run in runs_lower:
    df_lower = run.history()
    df_lower.columns = [run.name + c for c in df_lower.columns]
    l_dfs_lower.append(df_lower)
l_dfs_merged = l_dfs_mask + l_dfs_upper + l_dfs_lower
df_conditions = pd.concat(l_dfs_merged, axis=1).reset_index()
df_conditions.rename(columns={"index":"epoch"}, inplace=True)
df_conditions.to_csv("wandb/wandb_export_itc_masklength_code.csv", index=False)
```

## Data Set Size

```python
# load filtered runs from project

n_data = ["fewdata", "meddata", "muchdata"]
dataset_name = ["itc_small", "itc_medium", "itc_full"]

for ds_name, nd in zip(dataset_name, n_data):
    
    pth_save = f"""wandb/wandb_export_itc_conditions_{nd}_code.csv"""
    
    api = wandb.Api()
    
    runs_size = api.runs(
        "mirkothalmann-helmholtz-munich/source-of-variablity",
        filters={
            "config.dataset":  ds_name,
            "config.more_shuffle": "nothing",
            "config.rnd_seed": 1,
            "config.masktype": "causal"
        }
    )
    l_dfs_size = []
    for run in runs_size:
        df_size = run.history()
        df_size.columns = [run.name + c for c in df_size.columns]
        l_dfs_size.append(df_size)
    df_variables = pd.concat(l_dfs_size, axis=1).reset_index()
    df_variables.rename(columns={"index":"epoch"}, inplace=True)
    df_variables.to_csv(pth_save, index=False)
```
